```python
# 3. Εφαρμογή Αλγορίθμων
# 2 Classification Models:
#   1. Naive Bayes
#   2. Decision Tree
# Αρχικά, χωρίζω Χ = όλα τα χαρακτηριστικά εισόδου και Υ = η στήλη class που θέλουμε να προβλέψουμε
X = train_df.drop('class', axis=1)
y = train_df['class']

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Μετατροπή target σε αριθμούς
# anomaly : 0 , normal : 1
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

print("Classes:", label_encoder.classes_)
print("Encoded labels:", set(y_encoded))
```

```python
# Μετατροπη στηλών που είναι κείμενα σε αριθμητικές
X_encoded = pd.get_dummies(X, drop_first=False)

print("Shape before encoding:", X.shape)
print("Shape after encoding:", X_encoded.shape)
```

```python
# Το Test_data δεν έχει στηλη class, άρα δεν μπορούμε να το χρησιμοποιήσουμε για κανονική αξιολόγηση.
# Γι'αυτό χωρίζω το Train_data σε training και testing
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y_encoded,
    test_size=0.2,
    random_state=42,
    stratify=y_encoded
)
# Διαστάσεις δεδομένων (γραμμές, στήλες)
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
# Scaling για μοντέλα που το χρειάζονται, όπως το Naive Bayes
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Naive Bayes

from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

# Δημιουργία και train Naive Bayes model
nb_model = GaussianNB()
nb_model.fit(X_train_scaled, y_train)

# Προβλέψεις
y_pred_nb = nb_model.predict(X_test_scaled)
```

```python
# Decision Tree

from sklearn.tree import DecisionTreeClassifier

# Δημιουργία και train Decision Tree model
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)

y_pred_dt = dt_model.predict(X_test)
```

```python
# Advanced Technique: K-Means Clustering
from sklearn.cluster import KMeans

# Δημιουργώ K-Means model με 2 clusters
kmeans_model = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

kmeans_clusters = kmeans_model.fit_predict(X_train_scaled)

print("K-Means clustering completed successfully.")
print("Number of clusters:", len(set(kmeans_clusters)))
```

```python
# Επειδή εχω 118 χαρακτηριστικά, χρησιμοποιώ PCA
from sklearn.decomposition import PCA

pca = PCA(n_components=2, random_state=42)
X_train_pca = pca.fit_transform(X_train_scaled)

# Plot K-Means clusters
plt.figure(figsize=(8, 6))
plt.scatter(
    X_train_pca[:, 0],
    X_train_pca[:, 1],
    c=kmeans_clusters,
    s=10
)

plt.title("K-Means Clustering Visualization using PCA")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.show()
```

```python
# Model Evaluation
# The evaluation metrics used are:
# - Accuracy
# - Precision
# - Recall
# - F1-score
# - Confusion Matrix
# - ROC Curve
```

```python
# Naive Bayes Evaluation
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

# Βρίσκουμε ποια αριθμητική τιμή αντιστοιχεί στην κλάση anomaly
anomaly_label = list(label_encoder.classes_).index('anomaly')

# Evaluation metrics for Naive Bayes
nb_accuracy = accuracy_score(y_test, y_pred_nb)
nb_precision = precision_score(y_test, y_pred_nb, pos_label=anomaly_label)
nb_recall = recall_score(y_test, y_pred_nb, pos_label=anomaly_label)
nb_f1 = f1_score(y_test, y_pred_nb, pos_label=anomaly_label)

print("Naive Bayes Evaluation Results")
print("Accuracy:", nb_accuracy)
print("Precision:", nb_precision)
print("Recall:", nb_recall)
print("F1-score:", nb_f1)

print("\nClassification Report:")
print(classification_report(y_test, y_pred_nb, target_names=label_encoder.classes_))
```

```python
# Confusion Matrix - Naive Bayes
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_nb,
    display_labels=label_encoder.classes_,
    cmap="Blues"
)

plt.title("Confusion Matrix - Naive Bayes")
plt.show()
```

```python
# Decision Tree Evaluation

dt_accuracy = accuracy_score(y_test, y_pred_dt)
dt_precision = precision_score(y_test, y_pred_dt, pos_label=anomaly_label)
dt_recall = recall_score(y_test, y_pred_dt, pos_label=anomaly_label)
dt_f1 = f1_score(y_test, y_pred_dt, pos_label=anomaly_label)

print("Decision Tree Evaluation Results")
print("Accuracy:", dt_accuracy)
print("Precision:", dt_precision)
print("Recall:", dt_recall)
print("F1-score:", dt_f1)

print("\nClassification Report:")
print(classification_report(y_test, y_pred_dt, target_names=label_encoder.classes_))
```

```python
# Confusion Matrix — Decision Tree
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_dt,
    display_labels=label_encoder.classes_,
    cmap="Greens"
)

plt.title("Confusion Matrix - Decision Tree")
plt.show()
```

```python
# Επειδή και τα δύο μοντέλα υποστηρίζουν predict_proba(), μπορούμε να βγάλουμε ROC Curve.
from sklearn.metrics import roc_curve, auc

nb_scores = nb_model.predict_proba(X_test_scaled)[:, anomaly_label]
dt_scores = dt_model.predict_proba(X_test)[:, anomaly_label]

# ROC curve for Naive Bayes
fpr_nb, tpr_nb, _ = roc_curve(y_test, nb_scores, pos_label=anomaly_label)
roc_auc_nb = auc(fpr_nb, tpr_nb)

# ROC curve for Decision Tree
fpr_dt, tpr_dt, _ = roc_curve(y_test, dt_scores, pos_label=anomaly_label)
roc_auc_dt = auc(fpr_dt, tpr_dt)

plt.figure(figsize=(8, 6))

plt.plot(fpr_nb, tpr_nb, label=f"Naive Bayes (AUC = {roc_auc_nb:.4f})")
plt.plot(fpr_dt, tpr_dt, label=f"Decision Tree (AUC = {roc_auc_dt:.4f})")

plt.plot([0, 1], [0, 1], linestyle="--", label="Random Classifier")

plt.title("ROC Curve")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.show()
```

```python
# Συγκεντρωτικός πίνακας αξιολόγησης
model_results = pd.DataFrame({
    "Model": ["Naive Bayes", "Decision Tree"],
    "Accuracy": [nb_accuracy, dt_accuracy],
    "Precision": [nb_precision, dt_precision],
    "Recall": [nb_recall, dt_recall],
    "F1-score": [nb_f1, dt_f1],
    "ROC AUC": [roc_auc_nb, roc_auc_dt]
})

model_results
```

```python
# Γράφημα σύγκρισης metrics
model_results.set_index("Model").plot(kind="bar", figsize=(10, 6))

plt.title("Model Evaluation Results")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.legend(loc="lower right")
plt.show()
```
